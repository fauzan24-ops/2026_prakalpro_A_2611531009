Python merupakan salah satu bahasa pemrograman yang banyak digunakan karena memiliki syntax yang sederhana dan mudah dipahami, terutama bagi pemula yang sedang mempelajari dasar-dasar pemrograman. Dalam membuat sebuah program, pemahaman mengenai Operator yang ada dalam bahasa pemrograman python.

## 2.1 Dasar Teori

1. Operator Aritmatika.
digunakan untuk melakukan operasi matematika dasar seperti penjumlahan, pengurangan, perkalian, pembagian, dan sisa bagi (modulus), dan perpangkatan pada suatu nilai atau variabel.

2. Operator Perbandingan (Comparison).
digunakan untuk membandingkan dua buah nilai atau variabel, di mana hasil dari operasi ini berupa nilai kebenaran boolean yaitu True atau False.

3. Operator Logika.
digunakan untuk menggabungkan beberapa kondisi atau ekspresi boolean menggunakan logika tertentu seperti and, or, dan not.

4. Operator Bitwise.
digunakan untuk melakukan operasi manipulasi tingkat bit (biner) secara langsung pada bilangan bulat (integer).

5. Operator Penugasan (Assignment).
Operator penugasan digunakan untuk memberikan, mengisi, atau memperbarui nilai ke dalam suatu variabel di dalam program.

6. Variabel.
Variabel merupakan tempat untuk menyimpan data dalam program. Python dapat menentukan tipe data variabel berdasarkan nilai yang dimasukkan/ diinput.

7. Input dan Output.
Digunakan untuk menerima data dari pengguna dengan input(), sedangkan output digunakan untuk menampilkan hasil menggunakan print().
