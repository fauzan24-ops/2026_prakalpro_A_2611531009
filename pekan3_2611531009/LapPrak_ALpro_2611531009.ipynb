## 2.1 Dasar Teori

1. Tipe Data.
Tipe data menentukan jenis nilai yang digunakan dalam program. Tipe data dasar Python meliputi numerik untuk bilangan, string untuk teks, dan boolean untuk nilai true or false.

2. Variabel.
Variabel merupakan tempat untuk menyimpan data dalam program. Python dapat menentukan tipe data variabel berdasarkan nilai yang dimasukkan/ diinput.

3. Input dan Output.
Digunakan untuk menerima data dari pengguna dengan input(), sedangkan output digunakan untuk menampilkan hasil menggunakan print().

4. Konstanta.
Meruapakan nilai yang digunakan dalam program dan tidak diinginkan untuk diubah. Biasanya konstanta ditulis menggunakan huruf kapital, seperti PI = 3.14.
