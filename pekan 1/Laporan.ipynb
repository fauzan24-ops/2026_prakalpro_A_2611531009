 1.1 Latar Belakang
Python adalah bahasa pemrograman yang banyak digunakan karna syntax yang sederhana dan paling mendekati bahasa manusia, sehingga cocok untuk mempelajari dasar-dasar pemrograman. Memahami konsep dasar Python diperlukan untuk mempelajari materi yang lebih lanjut, seperti type and variabel, operator, statement conditional, perulangan for, while and do=while, string, and array.
Program "Hello World" merupakan contoh program data yang digunakan sederhana untuk menulis dan menjalankan program python. Selain itu, penggunaan fungsi print() untuk menampilkan karakter/data pada layar. Konsep ini dapat menjadi dasar dalam membuat program Python yang lebih rumit.

 2.1 Dasar Teori

1. Python dan Program Hello World.
Python dapat digunakan untuk membuat program dengan bahasa pemrogrman yang paling mndekati bahasa manusia. Program Hello World merupakan salah satu contoh sederhana untuk memulai pemrograman Python. Program ini menggunakan fungsi print() untuk menampilkan data atau teks ke layar melalui Visual Studio Code.
 
2. Menjalankan Program Python di VS Code. 
Program Python dapat dibuat dalam file dengan ekstensi .py. Setelah Python dan ekstension Python pada VS Code tersedia, file tersebut dapat dijalankan menggunakan tombol Run Python File(shift + enter). VSCode juga menyediakan pilihan untuk menjalankan baris atau bagian kode tertentu melalui Python Terminal.

3. Repository dan Remote Repository
Repository digunakan untuk menyimpan kode, file, dan riwayat perubahan suatu proyek. Repository dapat bersifat public atau private. Sedangkan, Remote repository merupakan repository yang berada di tempat lain, seperti GitHub, dan dapat dihubungkan dengan repository lokal menggunakan URL tertentu.
